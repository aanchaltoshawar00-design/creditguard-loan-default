```python
import pandas as pd

cols = ["loan_amnt", "term", "int_rate", "installment", "grade", "emp_length",
        "home_ownership", "annual_inc", "verification_status", "issue_d",
        "loan_status", "purpose", "addr_state", "dti", "delinq_2yrs",
        "fico_range_low", "fico_range_high", "open_acc", "pub_rec",
        "revol_bal", "revol_util", "total_acc"]

df = pd.read_csv("../data/raw/accepted_2007_to_2018Q4.csv.gz",
                 usecols=cols, low_memory=False)
print(df.shape)
print(df["loan_status"].value_counts())
```

```python
# Loans still "Current" have no outcome yet, so drop them
df = df[df["loan_status"].isin(["Fully Paid", "Charged Off"])]

df = df.sample(n=250_000, random_state=42)
df.to_csv("../data/processed/loans_sample.csv", index=False)

print(df.shape)
print(df["loan_status"].value_counts(normalize=True))
```
